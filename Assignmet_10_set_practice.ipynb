### Q27. Three-Set Analysis
- python = {"Ali", "Sara", "Ahmed", "Zain"}
- sql = {"Sara", "Ahmed", "Hamza"}
- aws = {"Ahmed", "Zain", "Hamza"}
- Find students in all three, exactly Python and SQL (not AWS), and at least two.

```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

students_in_all_three = python.intersection(sql).intersection(aws)
students_in_all_three
py_sql_not_in_aws = python.intersection(sql) - aws
py_sql_not_in_aws
py_sql = python.intersection(sql)
py_aws = python.intersection(aws)
aws_sql = aws.intersection(sql)
py_sql, py_aws, aws_sql
```

### Q28. Transaction Analysis
- day1 = {101, 102, 103, 104}
- day2 = {103, 104, 105, 106}
- day3 = {104, 106, 107, 108}
- Find IDs appearing on all days, exactly one day, and at least two days.

```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}
all_days = day1.intersection(day2).intersection(day3)
all_days
# at least 2 days: (AnB)U(AnC)U(BnC)
two_days = (day1.intersection(day2)).union(day1.intersection(day3)).union(day2.intersection(day3))
two_days
# exactly one day: (AuBuC) - ((AnB)U(AnC)U(BnC))
one_day_only = (day1.union(day2).union(day3))-two_days
one_day_only
```

### Q30. Customer Segmentation
- website = {"C1", "C2", "C3", "C4", "C5"}
- mobile = {"C3", "C4", "C6", "C7"}
- store = {"C2", "C4", "C7", "C8"}
- Find customers on exactly one, exactly two, and all three platforms.

```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website.intersection(mobile, store)
web_mob = website.intersection(mobile)
web_store = website.intersection(store)
mob_store = mobile.intersection(store)
exactly_two = web_mob.union(web_store, mob_store).difference(all_three)
exactly_one = website.symmetric_difference(mobile).symmetric_difference(store).difference(all_three)

print("Exactly one platform:", exactly_one)
print("Exactly two platforms:", exactly_two)
print("All three platforms:", all_three)
```
