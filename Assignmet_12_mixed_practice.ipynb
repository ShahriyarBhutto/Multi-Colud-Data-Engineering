### Q41. Student Performance System
- students = [
 {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
 {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
 {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]
- Calculate each average, top student, tuple of names, and set of unique marks.

```python
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

averages = {
    students[0]["name"]: sum(students[0]["marks"]) / len(students[0]["marks"]),
    students[1]["name"]: sum(students[1]["marks"]) / len(students[1]["marks"]),
    students[2]["name"]: sum(students[2]["marks"]) / len(students[2]["marks"])
}

top_student = max(averages, key=averages.get)
names = tuple(student["name"] for student in []) if False else (students[0]["name"], students[1]["name"], students[2]["name"])
unique_marks = set(students[0]["marks"]) | set(students[1]["marks"]) | set(students[2]["marks"])

print(averages)
print(top_student)
print(names)
print(unique_marks)
```

### Q42. Employee Skills
- employees = {
 "Ali": {"Python", "SQL", "AWS"},
 "Sara": {"Python", "Databricks", "SQL"},
 "Ahmed": {"AWS", "Snowflake", "SQL"}
}
- Find common skills, Ali-only skills, all team skills, and employees meeting Python + SQL requirements.

```python
employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

common_skills = employees["Ali"] & employees["Sara"] & employees["Ahmed"]
ali_only = employees["Ali"] - employees["Sara"] - employees["Ahmed"]
all_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]
qualified = (
    ("Ali", employees["Ali"]),
    ("Sara", employees["Sara"]),
    ("Ahmed", employees["Ahmed"])
)

print(common_skills)
print(ali_only)
print(all_skills)
print(tuple(name for name, skills in qualified if {"Python", "SQL"} <= skills))
```

### Q43. Sales Data
- sales = [
 ("Laptop", 120000, 2),
 ("Mouse", 1500, 5),
 ("Keyboard", 3500, 3)
]
- Calculate revenue, top-revenue product, name-to-revenue dictionary, and set of prices.

```python
sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

revenue = {
    sales[0][0]: sales[0][1] * sales[0][2],
    sales[1][0]: sales[1][1] * sales[1][2],
    sales[2][0]: sales[2][1] * sales[2][2]
}

top_product = max(revenue, key=revenue.get)
total_revenue = sum(revenue.values())
prices = {item[1] for item in sales}

print(total_revenue)
print(top_product)
print(revenue)
print(prices)
```

### Q45. Course Enrollment
- courses = {
 "Python": {"Ali", "Sara", "Ahmed"},
 "SQL": {"Sara", "Ahmed", "Zain"},
 "AWS": {"Ali", "Ahmed", "Zain"}
}
- Find students in exactly two, all three, and exactly one course; count unique students.

```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

python_students = courses["Python"]
sql_students = courses["SQL"]
aws_students = courses["AWS"]

all_three = python_students & sql_students & aws_students

exactly_two = (
    (python_students & sql_students)
    | (python_students & aws_students)
    | (sql_students & aws_students)
) - all_three

all_students = python_students | sql_students | aws_students
exactly_one = all_students - (
    (python_students & sql_students)
    | (python_students & aws_students)
    | (sql_students & aws_students)
)

print(exactly_two)
print(all_three)
print(exactly_one)
print(len(all_students))
```

### Q46. E-Commerce Transactions
- transactions = [
 {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
 {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
 {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]
- Calculate total revenue, Ali spending, largest transaction, and unique product names.

```python
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

totals = {
    transactions[0]["id"]: transactions[0]["items"][0][1] * transactions[0]["items"][0][2] + transactions[0]["items"][1][1] * transactions[0]["items"][1][2],
    transactions[1]["id"]: transactions[1]["items"][0][1] * transactions[1]["items"][0][2] + transactions[1]["items"][1][1] * transactions[1]["items"][1][2],
    transactions[2]["id"]: transactions[2]["items"][0][1] * transactions[2]["items"][0][2]
}

total_revenue = sum(totals.values())
ali_spending = totals[101] + totals[103]
largest_transaction = max(totals, key=totals.get)

products = (
    transactions[0]["items"][0][0],
    transactions[0]["items"][1][0],
    transactions[1]["items"][0][0],
    transactions[1]["items"][1][0],
    transactions[2]["items"][0][0]
)
unique_products = set(products)

print(total_revenue)
print(ali_spending)
print(largest_transaction)
print(unique_products)
```

### Q47. Pipeline Monitoring
- pipelines = {
 "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
 "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
 "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}
- Find total processed records, failed pipelines, largest pipeline total, and average records per run for each.

```python
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

totals = {
    "pipeline_1": sum(pipelines["pipeline_1"]["records"]),
    "pipeline_2": sum(pipelines["pipeline_2"]["records"]),
    "pipeline_3": sum(pipelines["pipeline_3"]["records"])
}

total_processed = sum(totals.values())
failed_pipelines = tuple(
    name for name, details in pipelines.items() if details["status"] == "failed"
)
largest_pipeline = max(totals, key=totals.get)

averages = {
    "pipeline_1": totals["pipeline_1"] / len(pipelines["pipeline_1"]["records"]),
    "pipeline_2": totals["pipeline_2"] / len(pipelines["pipeline_2"]["records"]),
    "pipeline_3": totals["pipeline_3"] / len(pipelines["pipeline_3"]["records"])
}

print(total_processed)
print(failed_pipelines)
print(largest_pipeline)
print(averages)
```

### Q48. Banking Transactions
- accounts = {
 "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
 "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
 "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}
- Assuming zero starting balances, calculate final balances, highest balance, total withdrawals, and largest single
deposit.

```python
accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

balances = {
    "A101": sum(accounts["A101"]["transactions"]),
    "A102": sum(accounts["A102"]["transactions"]),
    "A103": sum(accounts["A103"]["transactions"])
}

highest_balance = max(balances, key=balances.get)

total_withdrawals = (
    abs(min(accounts["A101"]["transactions"][1], 0))
    + abs(min(accounts["A101"]["transactions"][3], 0))
    + abs(min(accounts["A102"]["transactions"][1], 0))
    + abs(min(accounts["A102"]["transactions"][2], 0))
    + abs(min(accounts["A103"]["transactions"][1], 0))
)

largest_deposit = max(
    accounts["A101"]["transactions"][0],
    accounts["A101"]["transactions"][2],
    accounts["A102"]["transactions"][0],
    accounts["A103"]["transactions"][0],
    accounts["A103"]["transactions"][2]
)

print(balances)
print(highest_balance)
print(total_withdrawals)
print(largest_deposit)
```

### Q49. Product Recommendations
- users = {
 "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
 "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
 "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}
- Find products bought by at least two users, Ali wishlist minus purchases, shared wishlist products, and all products.

```python
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

bought_by_two = (
    (users["Ali"]["purchased"] & users["Sara"]["purchased"])
    | (users["Ali"]["purchased"] & users["Ahmed"]["purchased"])
    | (users["Sara"]["purchased"] & users["Ahmed"]["purchased"])
)

ali_wishlist_remaining = users["Ali"]["wishlist"] - users["Ali"]["purchased"]

shared_wishlist = (
    users["Ali"]["wishlist"]
    & users["Sara"]["wishlist"]
    & users["Ahmed"]["wishlist"]
)

all_products = (
    users["Ali"]["purchased"] | users["Ali"]["wishlist"]
    | users["Sara"]["purchased"] | users["Sara"]["wishlist"]
    | users["Ahmed"]["purchased"] | users["Ahmed"]["wishlist"]
)

print(bought_by_two)
print(ali_wishlist_remaining)
print(shared_wishlist)
print(all_products)
```

### Q50. Final Company Analysis
- company = {
 "departments": {
 "Engineering": {
 "employees": [
 ("Ali", 120000, {"Python", "AWS", "SQL"}),
 ("Sara", 150000, {"Python", "Databricks", "SQL"})
 ],
 "budget": 500000
 },
 "Analytics": {
 "employees": [
 ("Ahmed", 100000, {"SQL", "Power BI"}),
 ("Zain", 130000, {"Python", "SQL", "Tableau"})
 ],
 "budget": 400000
 }
 }
}
1. Payroll per department.
2. Company payroll.
3. Highest-paid employee.
4. Skills shared by all.
5. Engineering-only skills.
6. Remaining budget per department.
7. Department with higher average salary.
8. Name-to-salary dictionary.
9. All unique skills.
10. Is combined budget sufficient?

```python
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

engineering = company["departments"]["Engineering"]
analytics = company["departments"]["Analytics"]

engineering_payroll = engineering["employees"][0][1] + engineering["employees"][1][1]
analytics_payroll = analytics["employees"][0][1] + analytics["employees"][1][1]

company_payroll = engineering_payroll + analytics_payroll

salaries = {
    engineering["employees"][0][0]: engineering["employees"][0][1],
    engineering["employees"][1][0]: engineering["employees"][1][1],
    analytics["employees"][0][0]: analytics["employees"][0][1],
    analytics["employees"][1][0]: analytics["employees"][1][1]
}

highest_paid = max(salaries, key=salaries.get)

shared_skills = (
    engineering["employees"][0][2]
    & engineering["employees"][1][2]
    & analytics["employees"][0][2]
    & analytics["employees"][1][2]
)

engineering_skills = (
    engineering["employees"][0][2] | engineering["employees"][1][2]
) - (
    analytics["employees"][0][2] | analytics["employees"][1][2]
)

engineering_remaining = engineering["budget"] - engineering_payroll
analytics_remaining = analytics["budget"] - analytics_payroll

engineering_average = engineering_payroll / len(engineering["employees"])
analytics_average = analytics_payroll / len(analytics["employees"])

higher_average_salary = (
    "Engineering" if engineering_average > analytics_average else "Analytics"
)

all_skills = (
    engineering["employees"][0][2]
    | engineering["employees"][1][2]
    | analytics["employees"][0][2]
    | analytics["employees"][1][2]
)

budget_sufficient = (
    engineering["budget"] + analytics["budget"] >= company_payroll
)

print(engineering_payroll)
print(analytics_payroll)
print(company_payroll)
print(highest_paid)
print(shared_skills)
print(engineering_skills)
print({"Engineering": engineering_remaining, "Analytics": analytics_remaining})
print(higher_average_salary)
print(salaries)
print(all_skills)
print(budget_sufficient)
```
